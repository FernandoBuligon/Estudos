```python
import torch.nn as nn

class FeedForwardNetwork(nn.Module):
    def __init__(self, embedding_dim, ff_dim):
        super().__init__()

        self.layer1 = nn.Linear(embedding_dim, ff_dim) # primeira camada linear
        self.activation = nn.GELU() # função de ativação GeLU
        self.layer2 = nn.Linear(ff_dim, embedding_dim) # segunda camada linear

    def forward(self, x):
        x = self.layer1(x) # aplica a primeira camada linear
        x = self.activation(x) # aplica a função de ativação
        x = self.layer2(x) # aplica a segunda camada linear
        return x # retorna o resultado
```
