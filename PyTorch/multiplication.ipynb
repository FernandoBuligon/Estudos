Element-wise multiplication ('*'), 
Para conseguir multiplicar elemento por elementos, os dois tensor obrigatoriamente devem ter o mesmo formato

```python
import torch # importação da biblioteca PyTorch

a = torch.tensor([[1, 2], [3, 4]]) # criação de um tensor qualquer que servirá de exemplo
b = torch.tensor([[5, 6], [7, 8]]) # criação de outro tensor qualquer que servirá de exemplo

element_wise_product = a * b # multiplicação de a e b elemento a elemento

print(f"Tensor a: {a}") # imprime o tensor a
print(f"Tensor b: {b}") # imprime o tensor b
print(f"Multiplicação a * b: {element_wise_product}") # imprime o resultado da multiplicação elemento a elemento
```

Matrix multiplication ('@'), número de colunas da matriz 1 deve ser igual ao número de linhas da matriz 2

```python
m1 = torch.tensor([[1, 2, 3], [4, 5, 6]]) # criação de um tensor 2x3
m2 = torch.tensor([[7, 8], [9, 10], [11, 12]]) # criação de um tensor 3x2


matrix_product = m1 @ m2 # multiplicação de m1 e m2, resultado será um tensor 2x2

print(f"Tensor m1: {m1}") # imprime o tensor m1
print(f"Tensor m2: {m2}") # imprime o tensor m2
print(f"Multiplicação m1 @ m2: {matrix_product}") # imprime o resultado da multiplicação
```
