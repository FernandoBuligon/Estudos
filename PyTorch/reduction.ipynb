Reduction é qualquer operação que reduz o tensor em um número menor de elementos (sum, mean, max)

```python
import torch # importa a biblioteca PyTorch
scores = torch.tensor([[10., 20., 30.],[5., 10., 15.]])

average_score = scores.mean() # Soma todos os valores dos elementos e divide pelo numero de elementos total

print(f"Média geral: {average_score}")
max_score = scores.max() # Retorna o maior valor
min_score = scores.min() # Retorna o menor valor
print(f"Maior nota: {max_score}") 
print(f"Menor nota: {min_score}")
```

O argumento dim possibilita escolher qual direção colapsar

dim = 0 colapsa as linhas. dim = 1 colapsa as colunas

```python
avg_per_student = scores.mean(dim=1)  # colapsa as colunas, pegando a média de cada aluno
print(f"Média por aluno: {avg_per_student}")
avg_per_assignment = scores.mean(dim=0)  # colapsa as linhas, pegando a média de cada nota
print(f"Média por prova: {avg_per_assignment}")
```
