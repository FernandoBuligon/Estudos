```python
import torch # importação da biblioteca PyTorch

x_data = torch.tensor([[1, 2], [3, 4]]) # criação de um tensor qualquer que servirá de exemplo

w = torch.tensor([[1.0], [2.0]], requires_grad=True) # criação de um tensor de pesos com gradiente habilitado

print(f"Tensor: {x_data}") # imprime o tensor
print(f"data tensor requires grad: {x_data.requires_grad}") # imprime se o tensor de dados requer gradiente
print(f"weights tensor requires grad: {w.requires_grad}") # imprime se o tensor de pesos requer gradiente
```

Objetivo: Calcular z = x * y, em que y = a + b 

```python
a = torch.tensor(2.0, requires_grad=True) # criação de um tensor escalar com gradiente habilitado
b = torch.tensor(3.0, requires_grad=True) # criação de outro tensor escalar com gradiente habilitado
x = torch.tensor(4.0, requires_grad=True) # criação de mais um tensor escalar com gradiente habilitado

y = a + b # primeira operação
z = x * y # segunda operação

print(f"grad_fn for a: {a.grad_fn}") # a foi criado manualmente, então não tem grad_fn
print(f"grad_fn for y: {y.grad_fn}") # y foi criado por adição
print(f"grad_fn for z: {z.grad_fn}") # z foi criado por multiplicação
print(f"Resultado da primeira operação: y = {y}") # imprime o resultado da primeira operação
print(f"Resultado da segunda operação: z = {z}") # imprime o resultado da segunda operação
```
