```python
import torch # importação da biblioteca PyTorch

tensor = torch.tensor([[1,2],[3,4]]) # criação de um tensor qualquer que servirá de exemplo
print(f"Tensor: {tensor}") # imprime o tensor
print(f"Shape: {tensor.shape}") # imprime o shape do tensor
print(f"Dtype: {tensor.dtype}") # imprime o tipo de dado do tensor
print(f"Device: {tensor.device}") # imprime o dispositivo em que o tensor está alocado (CPU ou GPU)
```
