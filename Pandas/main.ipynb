```python
import pandas as pd
```

```python
# Series
values = [1, 2, 3, 4, 5] # lista de exemplo

s = pd.Series(values) # indexa a lista como uma Series do Pandas
print(f"Series:\n{s}") # imprime a Series

s_labeled = pd.Series(values, index=['a', 'b', 'c', 'd', 'a']) # cria uma Series com rótulos personalizados
print(f"\nSeries com rótulos:\n{s_labeled}") # imprime a Series com rótulos

elementos_a = s_labeled['a'] # acessa os elementos da Series com rótulo 'a'
print(f"\nElementos com rótulo 'a':\n{elementos_a}") # acessa os elementos da Series com rótulo 'a'
```

```python
# Data Frame

# indexa automaticamente
df = pd.DataFrame({
    'name': ['Fernando', 'Mille'],
    'age': [19, 3],
    'job': ['Estudante', 'cachorro']
})
print(f"\nDataFrame:\n{df}") # imprime o DataFrame

df = df.set_index('name') # define a coluna 'name' como índice do DataFrame
print(f"\nDataFrame com índice definido manualmente:\n{df}") # imprime o DataFrame com

fernando = df.loc['Fernando'] # acessa a linha do DataFrame com índice 'Fernando'
print(f"\nLinha com índice 'Fernando':\n{fernando}") # imprime a
```

```python
# Criação de dois dataframes para testar operações aritméticas

df1 = pd.DataFrame({
    'a': [1, 2, 3],
})

df2 = pd.DataFrame({
    'a': [1, 2, 3],
})

print(f"\nDataFrame 1:\n{df1}") # imprime o DataFrame 1
print(f"\nDataFrame 2:\n{df2}") # imprime o DataFrame 2
print(f"\nSoma dos DataFrames:\n{df1 + df2}") # imprime a soma dos DataFrames
print(f"\nSubtração dos DataFrames:\n{df1 - df2}") # imprime a subtração dos DataFrames
print(f"\nMultiplicação dos DataFrames:\n{df1 * df2}") # imprime a multiplicação dos DataFrames
print(f"\nDivisão dos DataFrames:\n{df1 / df2}") # imprime a divisão dos DataFrames
```

```python
# Criação de dois dataframes indexados manualmente para testar operações aritméticas 
# as operações são realizadas pelos indexes, não pela ordem das linhas 

df1 = pd.DataFrame({
    'a': [1, 2, 3]
},index=['a', 'b', 'c'])

df2 = pd.DataFrame({
    'a': [1, 2, 3],
},index=['c', 'b', 'a'])

print(f"\nDataFrame 1:\n{df1}") # imprime o DataFrame 1
print(f"\nDataFrame 2:\n{df2}") # imprime o DataFrame 2
print(f"\nSoma dos DataFrames:\n{df1 + df2}") # imprime a soma dos DataFrames
print(f"\nSubtração dos DataFrames:\n{df1 - df2}") # imprime a subtração dos DataFrames
print(f"\nMultiplicação dos DataFrames:\n{df1 * df2}") # imprime a multiplicação dos DataFrames
print(f"\nDivisão dos DataFrames:\n{df1 / df2}") # imprime a divisão dos DataFrames
```
